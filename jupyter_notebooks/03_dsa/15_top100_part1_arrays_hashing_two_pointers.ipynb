## How to solve a problem you have never seen

1. **Say it back, with an example.** Tell yourself the question in your own words, then make up a tiny example
   (4–6 items) and work out the answer **by hand**. If you cannot do it by hand, you cannot code it yet.
2. **Brute force, out loud.** Say the slow, obvious way — *"try every pair"* — and what it costs (O(n²)). It is a
   real answer, and it is the thing you will speed up.
3. **Find the repeated work.** Where does the slow way do the same job again and again? Looking for the same
   partner over and over? Adding up the same numbers again? That repeated job is what the trick removes.
4. **Pick the trick from the signal words.** Most questions whisper their trick. Use the table below.
5. **Test tiny and weird inputs.** Empty list, one item, all the same, negatives, zeros, the answer at the very
   start or the very end. Run them in your head (or in a cell) before you say "done".

## The tricks in this notebook, in pictures

Almost every question in this notebook is one of four ideas. Watch each one once before the problems — then, in
every problem, you will recognise the dance.

1. **A dictionary is a notebook.** Write down what you have seen, so later you can ask *"have I seen it?"* or
   *"how many?"* instantly instead of searching the whole row again.
2. **Two pointers walk toward each other.** One finger on each end of the row; each step, one or both move inward.
3. **A reader and a writer walk the same way.** The reader looks at every item; the writer only moves when we keep
   one. That rewrites a list in place.
4. **Running totals (prefix sums).** Add as you walk. Then the sum of any stretch is one subtraction.

```python
def notebook_movie(word):
    m = Movie("A dictionary is a notebook", "Count the letters of a word, looking at each letter once")
    count = {}
    m.step(f"How many of each letter does `{word}` have? The notebook starts empty.",
           arr(list(word)), hmap(count, "notebook  (letter → how many)"))
    for i, ch in enumerate(word):
        past = {k: "done" for k in range(i)}
        if ch in count:
            count[ch] += 1
            cap = f"`{ch}` is already in the notebook, so just add one: `{ch} → {count[ch]}`. No searching the row!"
            role = "look"
        else:
            count[ch] = 1
            cap = f"`{ch}` is new. Write a fresh line: `{ch} → 1`."
            role = "new"
        m.step(cap, arr(list(word), hi={**past, i: "look"}, ptr={"i": i}),
               hmap(count, "notebook  (letter → how many)", hi={ch: role}))
    m.step("Done in **one walk**. Now *how many a's?* is answered instantly: the notebook says `3`.",
           arr(list(word), hi={k: "done" for k in range(len(word))}),
           hmap(count, "notebook  (letter → how many)", hi={"a": "good"}))
    return count, m

counts, m = notebook_movie("banana")
assert counts == {"b": 1, "a": 3, "n": 2}
m.show()
```

```python
def ends_movie(cards):
    cards = list(cards)
    m = Movie("Two pointers walk toward each other", "Turn a row of cards back to front, without a second row")
    L, R = 0, len(cards) - 1
    m.step("`L` stands on the left end, `R` on the right end. We want the row back to front.",
           arr(cards, ptr={"L": L, "R": R}))
    while L < R:
        cards[L], cards[R] = cards[R], cards[L]
        finished = {k: "done" for k in range(len(cards)) if k < L or k > R}
        m.step(f"Swap the two end cards (`{cards[R]}` ↔ `{cards[L]}`), then **both** fingers step toward the middle.",
               arr(cards, hi={**finished, L: "swap", R: "swap"}, ptr={"L": L, "R": R}))
        L, R = L + 1, R - 1
    m.step("The fingers met in the middle. Every card was touched once: **O(n) time, no extra row**.",
           arr(cards, hi={k: "good" for k in range(len(cards))}, ptr={"L": L, "R": R}))
    return cards, m

row, m = ends_movie([1, 2, 3, 4, 5, 6, 7])
assert row == [7, 6, 5, 4, 3, 2, 1]
m.show()
```

```python
def reader_writer_movie(nums):
    nums = list(nums)
    m = Movie("A reader and a writer walk the same way", "Keep only the even numbers, packed at the front, in place")
    w = 0
    m.step("The **reader** `r` looks at every card. The **writer** `w` marks the next seat for a card we keep.",
           arr(nums, ptr={"r": 0, "w": 0}))
    for r in range(len(nums)):
        kept = {k: "good" for k in range(w)}
        skipped = {k: "done" for k in range(w, r)}
        if nums[r] % 2 == 0:
            nums[w] = nums[r]
            m.step(f"`{nums[r]}` is even: keep it. Copy it into seat `{w}`, and the writer moves on.",
                   arr(nums, hi={**skipped, r: "look", w: "new", **kept}, ptr={"r": r, "w": w}))
            w += 1
        else:
            m.step(f"`{nums[r]}` is odd: skip it. Only the reader moves; the writer waits.",
                   arr(nums, hi={**kept, **skipped, r: "bad"}, ptr={"r": r, "w": w}))
    m.step(f"The first **{w}** seats hold the keepers, in their old order. Whatever is after them we simply ignore.",
           arr(nums, hi={**{k: "good" for k in range(w)}, **{k: "done" for k in range(w, len(nums))}}, ptr={"w": w}))
    return nums[:w], m

kept, m = reader_writer_movie([3, 8, 5, 2, 6, 7, 4])
assert kept == [8, 2, 6, 4]
m.show()
```

```python
def prefix_movie(nums, a, b):
    m = Movie("Running totals (prefix sums)", "Add as you walk; then any stretch's sum is one subtraction")
    pre = [0]
    m.step("`pre[i]` = the sum of the first `i` numbers. It starts with `0`, meaning *nothing added yet*.",
           arr(nums, "nums"), arr(pre, "pre  (running total)"))
    for i, x in enumerate(nums):
        pre.append(pre[-1] + x)
        m.step(f"Add `{x}`: the running total goes {pre[-2]} + {x} = **{pre[-1]}**.",
               arr(nums, "nums", hi={**{k: "done" for k in range(i)}, i: "look"}), arr(pre, "pre  (running total)", hi={i + 1: "new"}))
    s = pre[b + 1] - pre[a]
    m.step(f"Sum of the blue stretch `nums[{a}..{b}]`? It is `pre[{b + 1}] − pre[{a}]` = {pre[b + 1]} − {pre[a]} = **{s}**. No re-adding!",
           arr(nums, "nums", win=(a, b)), arr(pre, "pre  (running total)", hi={a: "look", b + 1: "good"}))
    return s, m

nums = [3, 1, 4, 1, 5, 9]
s, m = prefix_movie(nums, 2, 4)
assert s == sum(nums[2:5]) == 10
m.show()
```

## 1 · Two Sum · [LeetCode 1](https://leetcode.com/problems/two-sum/) · 🟢 Easy

**🧸 Like you're five:** You have a row of number cards and a magic number, 9. Find the two cards that add up to 9. As you walk
        along the row you write every card you pass in a notebook, so for each new card you only have to ask:
        *"is the partner I need already in my notebook?"*

**The question:** Given a list `nums` and a number `target`, return the positions (indices) of the two numbers that add up to `target`. There is exactly one answer and you may not use the same position twice.

| Input | Output |
|---|---|
| `nums = [2, 7, 11, 15], target = 9` | `[0, 1]` |
| `nums = [3, 2, 4], target = 6` | `[1, 2]` |
| `nums = [3, 3], target = 6` | `[0, 1]` |

**🔎 How to spot it:** "find a **pair** that adds up to a target" and the list is **not sorted** → a dictionary from value to
        index. (If the list *is* sorted, use two pointers instead — problem 14.)

**💡 The trick (hash map):** For each number `x`, the partner you need is `target - x`. Keep a dictionary `seen` of every number
        you have already passed. If the partner is in `seen`, you are done; otherwise write `x` down and move on.
        Check **before** you store `x`, so a number never pairs with itself.

**⏱ Cost:** **O(n)** time (one pass, each dictionary lookup is O(1)) · **O(n)** space for `seen`. The brute force — try every pair — is O(n²) time.

```python
def two_sum(nums, target):
    seen = {}                                  # number -> the index where we saw it
    for i, x in enumerate(nums):
        need = target - x                      # the partner that would make the total
        if need in seen:                       # did we already walk past it?
            return [seen[need], i]
        seen[x] = i                            # remember this number for later
    return []                                  # the problem promises we never get here

assert two_sum([2, 7, 11, 15], 9) == [0, 1]
assert two_sum([3, 2, 4], 6) == [1, 2]
assert two_sum([3, 3], 6) == [0, 1]            # same value twice: we checked before storing
assert two_sum([-3, 4, 3, 90], 0) == [0, 2]    # negatives work too
print(two_sum([2, 7, 11, 15], 9))
```

```python
def two_sum_movie(nums, target):
    m = Movie("Two Sum", f"Which two cards add up to {target}?")
    seen = {}
    m.step(f"We want two cards that add up to **{target}**. Our notebook `seen` starts empty.",
           arr(nums), hmap(seen, "seen  (number → index)"), note(target=target))
    for i, x in enumerate(nums):
        need = target - x
        past = {k: "done" for k in range(i)}
        if need in seen:
            j = seen[need]
            m.step(f"Card `{x}` needs a partner of `{need}` — and `{need}` IS in the notebook, at index {j}! 🎉",
                   arr(nums, hi={**past, j: "good", i: "good"}, ptr={"i": i}),
                   hmap(seen, "seen  (number → index)", hi={need: "good"}), note(target=target, need=need))
            m.step(f"Answer: **[{j}, {i}]**, because {nums[j]} + {x} = {target}.",
                   arr(nums, hi={j: "good", i: "good"}))
            return [j, i], m
        seen[x] = i
        m.step(f"Card `{x}` needs a partner of `{need}`. Not in the notebook, so write `{x} → {i}` down and move on.",
               arr(nums, hi={**past, i: "look"}, ptr={"i": i}),
               hmap(seen, "seen  (number → index)", hi={x: "new"}), note(target=target, need=need))
    return [], m

nums, target = [4, 11, 7, 2, 15], 9
answer, m = two_sum_movie(nums, target)
assert answer == two_sum(nums, target) == [2, 3]
m.show()
```

## 2 · Contains Duplicate · [LeetCode 217](https://leetcode.com/problems/contains-duplicate/) · 🟢 Easy

**🧸 Like you're five:** You are handing out stickers and nobody may get two. Every time a child comes up, you check your list
        of names. If the name is already on the list — caught a duplicate!

**The question:** Return `True` if any value appears at least twice in `nums`, and `False` if every value is different.

| Input | Output |
|---|---|
| `nums = [1, 2, 3, 1]` | `True` |
| `nums = [1, 2, 3, 4]` | `False` |
| `nums = []` | `False` |

**🔎 How to spot it:** "have I **seen this before**?" → a `set`. Membership in a set is O(1) on average.

**💡 The trick (hash set):** Walk through the numbers with an empty set. If the number is already in the set, return `True`.
        Otherwise add it. If you reach the end, every number was new. (One-liner: `len(set(nums)) < len(nums)` — fine
        to mention, but the loop stops early at the first duplicate.)

**⏱ Cost:** **O(n)** time · **O(n)** space. Sorting first and comparing neighbours is O(n log n) time and O(1) extra space — the trade-off interviewers like to hear.

```python
def contains_duplicate(nums):
    seen = set()
    for x in nums:
        if x in seen:          # met this one before
            return True
        seen.add(x)
    return False

assert contains_duplicate([1, 2, 3, 1]) is True
assert contains_duplicate([1, 2, 3, 4]) is False
assert contains_duplicate([]) is False
assert contains_duplicate([1, 1, 1, 3, 3, 4, 3, 2, 4, 2]) is True
print(contains_duplicate([1, 2, 3, 1]))
```

```python
def contains_duplicate_movie(nums):
    m = Movie("Contains Duplicate", "Has any number shown up twice?")
    seen = set()
    m.step("Start with an empty bag called `seen`.", arr(nums), hmap(seen, "seen"))
    for i, x in enumerate(nums):
        past = {k: "done" for k in range(i)}
        if x in seen:
            first = nums.index(x)
            m.step(f"`{x}` is already in the bag — we saw it at index {first}. Duplicate found! ✅",
                   arr(nums, hi={**past, first: "good", i: "good"}, ptr={"i": i}), hmap(seen, "seen", hi={x: "good"}))
            return True, m
        seen.add(x)
        m.step(f"`{x}` is new. Drop it in the bag.", arr(nums, hi={**past, i: "look"}, ptr={"i": i}),
               hmap(seen, "seen", hi={x: "new"}))
    m.step("Reached the end and every number was new → `False`.", arr(nums, hi={k: "done" for k in range(len(nums))}),
           hmap(seen, "seen"))
    return False, m

nums = [5, 3, 8, 1, 3, 9]
answer, m = contains_duplicate_movie(nums)
assert answer == contains_duplicate(nums) is True
m.show()
```

## 3 · Valid Anagram · [LeetCode 242](https://leetcode.com/problems/valid-anagram/) · 🟢 Easy

**🧸 Like you're five:** You build a word out of letter tiles. Can you knock it down and build the second word with **exactly**
        the same tiles — none missing, none left over?

**The question:** Given two strings `s` and `t`, return `True` if `t` is an anagram of `s` (the same letters, each used the same number of times, in any order), otherwise `False`.

| Input | Output |
|---|---|
| `s = "anagram", t = "nagaram"` | `True` |
| `s = "rat", t = "car"` | `False` |
| `s = "a", t = "ab"` | `False` |

**🔎 How to spot it:** "same letters", "anagram", "rearrange", "how many of each" → **count** with a dictionary
        (or `collections.Counter`).

**💡 The trick (counting):** If the lengths differ, answer `False` straight away. Count every letter of `s` **up** in a dictionary,
        then count every letter of `t` **down**. If a letter of `t` finds its count already at 0, `t` has a tile that
        `s` does not → `False`. The one-liner `Counter(s) == Counter(t)` is fine to mention; `sorted(s) == sorted(t)`
        works too but costs O(n log n).

**⏱ Cost:** **O(n)** time (one look per letter) · **O(1)** extra space for lowercase letters (at most 26 keys). Sorting both strings is O(n log n).

```python
def is_anagram(s, t):
    if len(s) != len(t):                       # different lengths can never match
        return False
    count = {}
    for ch in s:                               # count the tiles of s up ...
        count[ch] = count.get(ch, 0) + 1
    for ch in t:                               # ... and use them up with the tiles of t
        if count.get(ch, 0) == 0:              # t needs a tile that s has run out of
            return False
        count[ch] -= 1
    return True

assert is_anagram("anagram", "nagaram") is True
assert is_anagram("rat", "car") is False
assert is_anagram("a", "ab") is False          # different lengths
assert is_anagram("", "") is True
assert is_anagram("aacc", "ccac") is False     # same letters, wrong amounts

import random
rng = random.Random(0)
for _ in range(300):                           # compare with the slow-but-obvious answer
    s = "".join(rng.choice("abc") for _ in range(rng.randint(0, 6)))
    t = "".join(rng.choice("abc") for _ in range(rng.randint(0, 6)))
    assert is_anagram(s, t) == (sorted(s) == sorted(t))
print(is_anagram("anagram", "nagaram"))
```

```python
def is_anagram_movie(s, t):
    m = Movie("Valid Anagram", f"Are '{s}' and '{t}' made of the same letter tiles?")
    S, T, count = list(s), list(t), {}
    m.step(f"Count the tiles of `{s}` **up**, then use them up with the tiles of `{t}`. Nobody may run short.",
           arr(S, "s"), arr(T, "t"), hmap(count, "letter → how many left"))
    if len(s) != len(t):
        m.step("The words have different lengths, so they can never match → **False**.", arr(S, "s"), arr(T, "t"))
        return False, m
    for i, ch in enumerate(s):
        count[ch] = count.get(ch, 0) + 1
        m.step(f"`{ch}` from `s`: one more `{ch}` tile → `{ch} → {count[ch]}`.",
               arr(S, "s", hi={**{k: "done" for k in range(i)}, i: "look"}, ptr={"i": i}), arr(T, "t"),
               hmap(count, "letter → how many left", hi={ch: "new"}))
    for j, ch in enumerate(t):
        used = {k: "done" for k in range(j)}
        if count.get(ch, 0) == 0:
            m.step(f"`{ch}` from `t`: there is no `{ch}` tile left! → **False**.",
                   arr(S, "s", hi={k: "done" for k in range(len(s))}), arr(T, "t", hi={**used, j: "bad"}, ptr={"j": j}),
                   hmap(count, "letter → how many left", hi={ch: "bad"}))
            return False, m
        count[ch] -= 1
        m.step(f"`{ch}` from `t`: use up one `{ch}` tile → `{ch} → {count[ch]}` left.",
               arr(S, "s", hi={k: "done" for k in range(len(s))}), arr(T, "t", hi={**used, j: "look"}, ptr={"j": j}),
               hmap(count, "letter → how many left", hi={ch: "look"}))
    m.step("Every tile of `t` was found and every count is back to `0` → **True**. 🎉",
           arr(S, "s", hi={k: "good" for k in range(len(s))}), arr(T, "t", hi={k: "good" for k in range(len(t))}),
           hmap(count, "letter → how many left", hi={c: "good" for c in count}))
    return True, m

answer, m = is_anagram_movie("apple", "papel")
assert answer == is_anagram("apple", "papel") is True
m.show()
```

## 4 · Group Anagrams · [LeetCode 49](https://leetcode.com/problems/group-anagrams/) · 🟡 Medium

**🧸 Like you're five:** You have a pile of word cards. Give every card a **sticker** with its letters in ABC order: `eat`,
        `tea` and `ate` all get the sticker `aet`. Now put cards with the same sticker in the same box.

**The question:** Given a list of strings `strs`, group the anagrams together. Return the groups in any order.

| Input | Output |
|---|---|
| `strs = ["eat","tea","tan","ate","nat","bat"]` | `[["bat"],["nat","tan"],["ate","eat","tea"]]` |
| `strs = [""]` | `[[""]]` |
| `strs = ["a"]` | `[["a"]]` |

**🔎 How to spot it:** "**group** things that are the same in some way" → a dictionary from a **key that is equal for all
        members of a group** to the list of members. The whole question is: what is the key?

**💡 The trick (hash map with a smart key):** The key of a word is its letters sorted: `"".join(sorted(w))`. Anagrams share the key, other words do
        not. Use `defaultdict(list)` and append each word to `groups[key]`. The trap: a dictionary key must be
        hashable, so use a string or a tuple — never a list. (A faster key is a tuple of 26 letter counts: O(k) instead
        of O(k log k) per word.)

**⏱ Cost:** **O(n · k log k)** time for n words of length up to k (sorting each word) · **O(n · k)** space for the groups. Comparing every pair of words is O(n² · k).

```python
from collections import defaultdict

def group_anagrams(strs):
    groups = defaultdict(list)                 # sorted letters -> the words that use them
    for w in strs:
        key = "".join(sorted(w))               # 'eat', 'tea' and 'ate' all become 'aet'
        groups[key].append(w)
    return list(groups.values())

def canon(groups):
    # Groups may come in any order: sort inside and outside so two answers can be compared.
    return sorted(sorted(g) for g in groups)

assert canon(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])) == [["ate", "eat", "tea"], ["bat"], ["nat", "tan"]]
assert group_anagrams([""]) == [[""]]
assert group_anagrams(["a"]) == [["a"]]
assert canon(group_anagrams(["ab", "ba", "ab"])) == [["ab", "ab", "ba"]]    # duplicates stay
print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
```

```python
def group_anagrams_movie(strs):
    m = Movie("Group Anagrams", "Words made of the same letters go in the same box")
    groups = {}

    def boxes():                                # how the dictionary looks: sticker -> the words in its box
        return {k: ", ".join(v) for k, v in groups.items()}

    m.step("Each word gets a **sticker**: its letters in ABC order. Same sticker → same box.",
           arr(strs), hmap(boxes(), "sticker → box"))
    for i, w in enumerate(strs):
        key = "".join(sorted(w))
        is_new = key not in groups
        groups.setdefault(key, []).append(w)
        if is_new:
            cap = f"`{w}` sorted is `{key}`. No box has that sticker yet, so open a **new box**."
        else:
            cap = f"`{w}` sorted is `{key}`. That box already exists, so `{w}` joins `{groups[key][0]}`."
        m.step(cap, arr(strs, hi={**{k: "done" for k in range(i)}, i: "look"}, ptr={"i": i}),
               hmap(boxes(), "sticker → box", hi={key: "new" if is_new else "look"}))
    answer = list(groups.values())
    m.step(f"Every word is in a box: **{len(answer)} groups**. We looked at each word only once.",
           arr(strs, hi={k: "done" for k in range(len(strs))}),
           hmap(boxes(), "sticker → box", hi={k: "good" for k in groups}))
    return answer, m

strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
answer, m = group_anagrams_movie(strs)
assert canon(answer) == canon(group_anagrams(strs))
m.show()
```

## 5 · Top K Frequent Elements · [LeetCode 347](https://leetcode.com/problems/top-k-frequent-elements/) · 🟡 Medium

**🧸 Like you're five:** Kids vote for their favourite fruit. First you count the votes. Then you make shelves numbered 1, 2,
        3… and put each fruit on the shelf that matches its votes. To find the winners, read the shelves from the top
        down.

**The question:** Given `nums` and `k`, return the `k` most frequent values, in any order. The answer is unique. Aim for better than O(n log n).

| Input | Output |
|---|---|
| `nums = [1,1,1,2,2,3], k = 2` | `[1, 2]` |
| `nums = [1], k = 1` | `[1]` |
| `nums = [4,4,-1,-1,-1,2], k = 2` | `[-1, 4]` |

**🔎 How to spot it:** "**top k / k most frequent / k most common**" → count with a dictionary, then either a heap of size k
        (O(n log k)) or **bucket sort by count** (O(n)), because a count can never be bigger than n.

**💡 The trick (count, then buckets):** 1. Count each value. 2. Make `n + 1` buckets; bucket `f` holds the values seen exactly `f` times.
        3. Walk the buckets from `n` down to `1`, collecting values until you have `k`. The common mistake is sorting
        the whole count dictionary — correct, but O(n log n), which the question asks you to beat.

**⏱ Cost:** **O(n)** time (count, fill buckets, one walk) · **O(n)** space. Sorting by count is O(n log n); `heapq.nlargest(k, …)` is O(n log k).

```python
def top_k_frequent(nums, k):
    count = {}
    for x in nums:
        count[x] = count.get(x, 0) + 1
    buckets = [[] for _ in range(len(nums) + 1)]   # buckets[f] = values seen exactly f times
    for x, f in count.items():
        buckets[f].append(x)
    out = []
    for f in range(len(nums), 0, -1):              # most frequent first
        for x in buckets[f]:
            out.append(x)
            if len(out) == k:
                return out
    return out

assert sorted(top_k_frequent([1, 1, 1, 2, 2, 3], 2)) == [1, 2]
assert top_k_frequent([1], 1) == [1]
assert sorted(top_k_frequent([4, 4, -1, -1, -1, 2], 2)) == [-1, 4]

import random
from collections import Counter
rng = random.Random(0)
for _ in range(300):                               # the k we return must have the k biggest counts
    nums = [rng.randint(-3, 3) for _ in range(rng.randint(1, 12))]
    c = Counter(nums)
    k = rng.randint(1, len(c))
    got = top_k_frequent(nums, k)
    assert len(set(got)) == k
    assert sorted((c[x] for x in got), reverse=True) == sorted(c.values(), reverse=True)[:k]
print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))
```

```python
def top_k_movie(nums, k):
    m = Movie("Top K Frequent Elements", f"Which {k} numbers show up most often?")
    count = {}
    m.step("First **count** every number. Then put each number on the shelf (bucket) that matches its count.",
           arr(nums), hmap(count, "number → times"))
    for i, x in enumerate(nums):
        count[x] = count.get(x, 0) + 1
        m.step(f"`{x}`: seen **{count[x]}** time{'s' if count[x] > 1 else ''} so far.",
               arr(nums, hi={**{j: "done" for j in range(i)}, i: "look"}, ptr={"i": i}),
               hmap(count, "number → times", hi={x: "new" if count[x] == 1 else "look"}))
    buckets = [[] for _ in range(len(nums) + 1)]

    def shelves():                                  # each bucket drawn as one box
        return [" ".join(map(str, b)) if b else None for b in buckets]

    title = "buckets  (box f = seen f times)"
    for x, f in count.items():
        buckets[f].append(x)
        m.step(f"`{x}` was seen {f} time{'s' if f > 1 else ''}, so it goes into bucket **{f}**.",
               hmap(count, "number → times", hi={x: "look"}), arr(shelves(), title, hi={f: "new"}))
    out = []
    for f in range(len(nums), 0, -1):
        for x in buckets[f]:
            out.append(x)
            read = {g: "done" for g in range(f + 1, len(nums) + 1)}
            if len(out) < k:
                m.step(f"Read the buckets from the **top** down. Bucket {f} gives `{x}`. That is {len(out)} of {k}.",
                       arr(shelves(), title, hi={**read, f: "look"}), note(answer=out))
            else:
                m.step(f"Bucket {f} gives `{x}` and now we have **{k}**: `{out}`. The bucket numbers did the sorting for us.",
                       arr(shelves(), title, hi={**read, f: "good"}), note(answer=out))
                return out, m
    return out, m

nums, k = [5, 3, 5, 1, 3, 5, 7], 2
answer, m = top_k_movie(nums, k)
assert answer == top_k_frequent(nums, k) == [5, 3]
m.show()
```

## 6 · Product of Array Except Self · [LeetCode 238](https://leetcode.com/problems/product-of-array-except-self/) · 🟡 Medium

**🧸 Like you're five:** Each child in a line wants to know: *"if I step out, what do all the others multiply to?"* One helper
        walks from the left end telling each child the product of everyone on their left. A second helper walks from
        the right end and multiplies in everyone on their right.

**The question:** Given `nums`, return `answer` where `answer[i]` is the product of every number except `nums[i]`. Run in O(n) time **without division**.

| Input | Output |
|---|---|
| `nums = [1, 2, 3, 4]` | `[24, 12, 8, 6]` |
| `nums = [-1, 1, 0, -3, 3]` | `[0, 0, 9, 0, 0]` |
| `nums = [2, 3]` | `[3, 2]` |

**🔎 How to spot it:** "everything **except me**", "**without division**", "product/sum of all the others" → the answer for
        `i` = (something from the **left** of `i`) × (something from the **right** of `i`): two passes.

**💡 The trick (left pass and right pass):** Pass 1, left to right: `answer[i]` = product of everything **left** of `i` (carry it in `left`).
        Pass 2, right to left: multiply `answer[i]` by the product of everything **right** of `i` (carry it in
        `right`). Store `answer[i]` **before** you multiply `nums[i]` into the carry — that is how `i` leaves itself
        out. Division would break on a zero, which is why the question bans it.

**⏱ Cost:** **O(n)** time (two walks) · **O(1)** extra space (the output list does not count). The brute force — multiply all the others for each `i` — is O(n²).

```python
def product_except_self(nums):
    n = len(nums)
    answer = [1] * n
    left = 1
    for i in range(n):                  # answer[i] = product of everything LEFT of i
        answer[i] = left
        left *= nums[i]                 # only now include nums[i], ready for the next box
    right = 1
    for i in range(n - 1, -1, -1):      # multiply in the product of everything RIGHT of i
        answer[i] *= right
        right *= nums[i]
    return answer

assert product_except_self([1, 2, 3, 4]) == [24, 12, 8, 6]
assert product_except_self([-1, 1, 0, -3, 3]) == [0, 0, 9, 0, 0]   # one zero: no division needed
assert product_except_self([0, 0, 2]) == [0, 0, 0]                 # two zeros
assert product_except_self([2, 3]) == [3, 2]

import math, random
rng = random.Random(0)
for _ in range(300):
    nums = [rng.randint(-3, 3) for _ in range(rng.randint(2, 8))]
    assert product_except_self(nums) == [math.prod(nums[:i] + nums[i + 1:]) for i in range(len(nums))]
print(product_except_self([1, 2, 3, 4]))
```

```python
def product_movie(nums):
    m = Movie("Product of Array Except Self", "Each box gets the product of all the OTHER numbers — no division")
    n = len(nums)
    answer = [1] * n
    m.step("Pass 1 walks **left → right** carrying `left` = the product of everything already passed.",
           arr(nums, "nums"), arr(answer, "answer"), note(left=1))
    left = 1
    for i in range(n):
        answer[i] = left
        if i == 0:
            cap = f"Nothing is left of index 0, so `answer[0] = 1`. Then `left` becomes 1 × {nums[0]} = {nums[0]}."
        else:
            cap = f"The blue numbers multiply to `{left}`, so `answer[{i}] = {left}`. Then `left` × {nums[i]} = {left * nums[i]}."
        m.step(cap, arr(nums, "nums", hi={**{k: "win" for k in range(i)}, i: "look"}, ptr={"i": i}),
               arr(answer, "answer", hi={i: "new"}), note(left=left))
        left *= nums[i]
    m.step("Each box now holds its **left** side. Pass 2 walks **right → left** carrying `right`.",
           arr(nums, "nums"), arr(answer, "answer"), note(right=1))
    right = 1
    for i in range(n - 1, -1, -1):
        before = answer[i]
        answer[i] *= right
        m.step(f"The blue numbers on the right multiply to `{right}`, so `answer[{i}]` = {before} × {right} = **{answer[i]}**.",
               arr(nums, "nums", hi={**{k: "win" for k in range(i + 1, n)}, i: "look"}, ptr={"i": i}),
               arr(answer, "answer", hi={**{k: "good" for k in range(i + 1, n)}, i: "new"}), note(right=right))
        right *= nums[i]
    m.step(f"Done: every box = (left side) × (right side) = **{answer}**, and we never divided.",
           arr(nums, "nums"), arr(answer, "answer", hi={k: "good" for k in range(n)}))
    return answer, m

nums = [2, 3, 4, 5]
answer, m = product_movie(nums)
assert answer == product_except_self(nums) == [60, 40, 30, 24]
m.show()
```

## 7 · Longest Consecutive Sequence · [LeetCode 128](https://leetcode.com/problems/longest-consecutive-sequence/) · 🟡 Medium

**🧸 Like you're five:** Number blocks are scattered on the floor. You want the longest staircase 1, 2, 3, 4… you can build.
        Only start counting at a block that has **no** block one smaller — that is the bottom step of a staircase.

**The question:** Given an unsorted list `nums`, return the length of the longest run of consecutive values (like 1, 2, 3, 4). It must run in **O(n)** time.

| Input | Output |
|---|---|
| `nums = [100, 4, 200, 1, 3, 2]` | `4` |
| `nums = [0,3,7,2,5,8,4,6,0,1]` | `9` |
| `nums = []` | `0` |

**🔎 How to spot it:** "**consecutive**" + "**unsorted**" + "**O(n)**" → sorting is banned (O(n log n)), so put everything
        in a **set** and ask *"is x − 1 here? is x + 1 here?"* in O(1).

**💡 The trick (set, count from the start of each run):** Put all numbers in a set. For each `x` in the set, if `x - 1` is **also** in the set, skip it — `x`
        is in the middle of a run and will be counted from the run's start. Otherwise `x` starts a run: count up
        `x+1, x+2, …` while they are in the set. Every number is counted up at most once, so it is O(n). The trap:
        looping over `nums` instead of the set, so a start that appears many times gets counted many times.

**⏱ Cost:** **O(n)** time (each number is walked over at most twice) · **O(n)** space for the set. Sorting first is O(n log n).

```python
def longest_consecutive(nums):
    s = set(nums)
    best = 0
    for x in s:
        if x - 1 in s:                 # not the bottom of a staircase: it gets counted from the bottom
            continue
        y = x
        while y + 1 in s:              # climb the staircase x, x+1, x+2, ...
            y += 1
        best = max(best, y - x + 1)
    return best

assert longest_consecutive([100, 4, 200, 1, 3, 2]) == 4
assert longest_consecutive([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]) == 9
assert longest_consecutive([]) == 0
assert longest_consecutive([1, 0, 1, 2]) == 3            # duplicates do not break the run
assert longest_consecutive([-2, -1, 5]) == 2              # negatives

import random
def brute(nums):                                          # sort, then scan for runs
    v, best, run = sorted(set(nums)), 0, 0
    for i, x in enumerate(v):
        run = run + 1 if i and x == v[i - 1] + 1 else 1
        best = max(best, run)
    return best
rng = random.Random(0)
for _ in range(300):
    nums = [rng.randint(-5, 10) for _ in range(rng.randint(0, 10))]
    assert longest_consecutive(nums) == brute(nums)
print(longest_consecutive([100, 4, 200, 1, 3, 2]))
```

```python
def longest_consecutive_movie(nums):
    m = Movie("Longest Consecutive Sequence", "The longest staircase like 1, 2, 3, 4 — the blocks are mixed up")
    s = set(nums)
    best, best_from = 0, None
    m.step("Put every block in a **set**, so asking *is 5 here?* is instant. Then look at each number once.",
           arr(nums), hmap(s, "the set"), note(best=best))
    for x in dict.fromkeys(nums):              # the set's numbers, in the order we first met them
        i = nums.index(x)
        if x - 1 in s:
            m.step(f"`{x}`: `{x - 1}` is in the set, so `{x}` is **not** a bottom step. Skip it — its staircase is counted from the bottom.",
                   arr(nums, hi={i: "bad"}, ptr={"x": i}), hmap(s, "the set", hi={x: "bad", x - 1: "look"}), note(best=best))
            continue
        y = x
        m.step(f"`{x}`: `{x - 1}` is **not** in the set, so a staircase starts here. Climb up from `{x}`.",
               arr(nums, hi={i: "look"}, ptr={"x": i}), hmap(s, "the set", hi={x: "look"}), note(run=1, best=best))
        while y + 1 in s:
            y += 1
            steps = range(x, y + 1)
            m.step(f"`{y}` is in the set too → the staircase is `{x}` to `{y}`, **{y - x + 1}** steps.",
                   arr(nums, hi={nums.index(v): "good" for v in steps}), hmap(s, "the set", hi={v: "good" for v in steps}),
                   note(run=y - x + 1, best=best))
        if y - x + 1 > best:
            best, best_from = y - x + 1, x
        m.step(f"`{y + 1}` is missing, so this staircase has **{y - x + 1}** step{'s' if y > x else ''}. Best so far: **{best}**.",
               arr(nums, hi={nums.index(v): "done" for v in range(x, y + 1)}),
               hmap(s, "the set", hi={v: "done" for v in range(x, y + 1)}), note(run=y - x + 1, best=best))
    m.step(f"Every number was looked at, and each staircase climbed only once → the answer is **{best}**.",
           arr(nums, hi={k: "good" for k, v in enumerate(nums) if best_from <= v < best_from + best}), note(best=best))
    return best, m

nums = [100, 4, 200, 1, 3, 2]
answer, m = longest_consecutive_movie(nums)
assert answer == longest_consecutive(nums) == 4
m.show()
```

## 8 · Majority Element · [LeetCode 169](https://leetcode.com/problems/majority-element/) · 🟢 Easy

**🧸 Like you're five:** A class votes for a pet. Pair up kids who voted for **different** pets and send each pair out of the
        room. If one pet got more than half the votes, some of its fans are always left standing at the end.

**The question:** Given `nums` of length n, return the value that appears **more than n / 2 times**. Such a value always exists. Follow-up: O(n) time and O(1) extra space.

| Input | Output |
|---|---|
| `nums = [3, 2, 3]` | `3` |
| `nums = [2, 2, 1, 1, 1, 2, 2]` | `2` |
| `nums = [7]` | `7` |

**🔎 How to spot it:** "appears **more than half** the time" + "**O(1) space**" → the voting trick. (Without the space rule, a
        counting dictionary is the easy answer.)

**💡 The trick (voting (boyer–moore)):** Keep one `candidate` and a `count`. For each number: if `count` is 0, it becomes the new candidate.
        Then a vote for the candidate adds 1, a vote for anyone else takes 1 away (one fan of each leaves the room).
        The majority has more fans than everyone else put together, so it is the candidate left at the end. This only
        works because a majority is **guaranteed** — otherwise you would need a second pass to check.

**⏱ Cost:** **O(n)** time · **O(1)** space. A counting dictionary is O(n) space; sorting and taking the middle is O(n log n).

```python
def majority_element(nums):
    candidate, count = None, 0
    for x in nums:
        if count == 0:                 # nobody is ahead: x becomes the new candidate
            candidate = x
        count += 1 if x == candidate else -1   # a fan adds one, anyone else cancels one out
    return candidate

assert majority_element([3, 2, 3]) == 3
assert majority_element([2, 2, 1, 1, 1, 2, 2]) == 2
assert majority_element([7]) == 7
assert majority_element([-1, -1, 5]) == -1

import random
rng = random.Random(0)
for _ in range(300):                   # build lists that really have a majority, then shuffle
    n = rng.randint(1, 11)
    major = rng.randint(-5, 5)
    nums = [major] * (n // 2 + 1) + [rng.randint(-5, 5) for _ in range(n - n // 2 - 1)]
    rng.shuffle(nums)
    assert majority_element(nums) == major
print(majority_element([2, 2, 1, 1, 1, 2, 2]))
```

```python
def majority_movie(nums):
    m = Movie("Majority Element", "Which number has more than half of the votes?")
    candidate, count = None, 0
    m.step("Keep one **candidate** and a **count**. A vote for it: count up. Any other vote: count down (they cancel).",
           arr(nums), note(candidate=candidate, count=count))
    for i, x in enumerate(nums):
        past = {k: "done" for k in range(i)}
        start = ""
        if count == 0:
            candidate = x
            start = f"Count is 0, so `{x}` becomes the new candidate. "
        if x == candidate:
            count += 1
            cap, role = start + f"`{x}` votes for the candidate → count up to **{count}**.", "good"
        else:
            count -= 1
            cap, role = f"`{x}` votes against `{candidate}` → one fan of each leaves; count down to **{count}**.", "bad"
        m.step(cap, arr(nums, hi={**past, i: role}, ptr={"i": i}), note(candidate=candidate, count=count))
    m.step(f"The candidate still standing is **{candidate}**. A real majority can never be cancelled out completely.",
           arr(nums, hi={k: "good" for k, v in enumerate(nums) if v == candidate}), note(candidate=candidate, count=count))
    return candidate, m

nums = [2, 2, 1, 1, 1, 2, 2]
answer, m = majority_movie(nums)
assert answer == majority_element(nums) == 2
m.show()
```

## 9 · Merge Sorted Array · [LeetCode 88](https://leetcode.com/problems/merge-sorted-array/) · 🟢 Easy

**🧸 Like you're five:** Two lines of kids are sorted by height, and the first line has empty spots at the back. To make one
        sorted line, fill the **last** spot first with the tallest kid from either line — then nobody gets pushed out of
        their place.

**The question:** `nums1` has `m + n` slots: its first `m` values are sorted, the last `n` are zeros. `nums2` has `n` sorted values. Merge `nums2` into `nums1` **in place** so all `m + n` values are sorted. Return nothing.

| Input | Output |
|---|---|
| `nums1 = [1,2,3,0,0,0], m = 3, nums2 = [2,5,6], n = 3` | `nums1 = [1,2,2,3,5,6]` |
| `nums1 = [1], m = 1, nums2 = [], n = 0` | `nums1 = [1]` |
| `nums1 = [0], m = 0, nums2 = [1], n = 1` | `nums1 = [1]` |

**🔎 How to spot it:** "merge two **sorted** lists **in place**" and the free space is at the **end** → fill from the back
        with the bigger of the two last values.

**💡 The trick (three pointers from the back):** Put pointer `i` on the last real value of `nums1`, `j` on the last value of `nums2`, and `w` on the
        last slot. Copy the bigger of `nums1[i]` and `nums2[j]` into `w`, and step that pointer and `w` back. Stop when
        `nums2` is used up — whatever is left at the front of `nums1` is already in place. Filling from the **front**
        would overwrite values of `nums1` you still need; that is the trap.

**⏱ Cost:** **O(m + n)** time (each value is moved once) · **O(1)** extra space. Copying both into a new list and sorting is O((m + n) log(m + n)) and uses extra space.

```python
def merge(nums1, m, nums2, n):
    i, j, w = m - 1, n - 1, m + n - 1   # last real value of nums1, last of nums2, the slot to fill
    while j >= 0:                        # when nums2 is used up, the rest of nums1 is already in place
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[w] = nums1[i]
            i -= 1
        else:
            nums1[w] = nums2[j]
            j -= 1
        w -= 1

def merged(nums1, m, nums2, n):          # small helper for the tests: merge a copy and return it
    nums1 = list(nums1)
    merge(nums1, m, nums2, n)
    return nums1

assert merged([1, 2, 3, 0, 0, 0], 3, [2, 5, 6], 3) == [1, 2, 2, 3, 5, 6]
assert merged([1], 1, [], 0) == [1]
assert merged([0], 0, [1], 1) == [1]                   # nums1 starts with nothing real
assert merged([4, 5, 6, 0, 0, 0], 3, [1, 2, 3], 3) == [1, 2, 3, 4, 5, 6]

import random
rng = random.Random(0)
for _ in range(300):
    a = sorted(rng.randint(-9, 9) for _ in range(rng.randint(0, 6)))
    b = sorted(rng.randint(-9, 9) for _ in range(rng.randint(0, 6)))
    assert merged(a + [0] * len(b), len(a), b, len(b)) == sorted(a + b)
print(merged([1, 2, 3, 0, 0, 0], 3, [2, 5, 6], 3))
```

```python
def merge_movie(nums1, m, nums2, n):
    nums1 = list(nums1)
    mv = Movie("Merge Sorted Array", "Pour nums2 into the empty slots at the back of nums1")
    i, j, w = m - 1, n - 1, m + n - 1

    def pic(hi1=None, hi2=None):
        return (arr(nums1, "nums1", hi=hi1, ptr={"i": i if i >= 0 else None, "w": w if w >= 0 else None}),
                arr(nums2, "nums2", hi=hi2, ptr={"j": j if j >= 0 else None}))

    mv.step("Fill from the **back**: the biggest value goes into the last slot, so nothing we still need gets overwritten.",
            *pic(hi1={k: "done" for k in range(m, m + n)}))
    while j >= 0:
        filled = {k: "good" for k in range(w + 1, m + n)}
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[w] = nums1[i]
            mv.step(f"`{nums1[i]}` (nums1) is bigger than `{nums2[j]}` (nums2), so `{nums1[i]}` moves back to slot {w}.",
                    *pic(hi1={**filled, i: "swap", w: "new"}, hi2={j: "look"}))
            i -= 1
        else:
            nums1[w] = nums2[j]
            if i >= 0:
                cap = f"`{nums2[j]}` (nums2) is not smaller than `{nums1[i]}` (nums1), so `{nums2[j]}` goes into slot {w}."
            else:
                cap = f"nums1 has nothing left to compare, so `{nums2[j]}` goes into slot {w}."
            mv.step(cap, *pic(hi1={**filled, **({i: "look"} if i >= 0 else {}), w: "new"}, hi2={j: "swap"}))
            j -= 1
        w -= 1
    mv.step("nums2 is used up. The values still at the front of nums1 were already in the right place → **done**.",
            *pic(hi1={k: "good" for k in range(m + n)}, hi2={k: "done" for k in range(n)}))
    return nums1, mv

answer, mv = merge_movie([1, 4, 7, 0, 0, 0], 3, [2, 3, 9], 3)
assert answer == merged([1, 4, 7, 0, 0, 0], 3, [2, 3, 9], 3) == [1, 2, 3, 4, 7, 9]
mv.show()
```

## 10 · Remove Duplicates from Sorted Array · [LeetCode 26](https://leetcode.com/problems/remove-duplicates-from-sorted-array/) · 🟢 Easy

**🧸 Like you're five:** Sorted crayons lie in a row, and some colours repeat. You walk along (the reader) and a friend (the
        writer) builds a new row at the front out of only the **first** crayon of each colour.

**The question:** `nums` is sorted. Remove the duplicates **in place** so each value appears once, keeping the order. Return `k`, the number of unique values; the first `k` slots of `nums` must hold them. What is after slot `k` does not matter.

| Input | Output |
|---|---|
| `nums = [1, 1, 2]` | `2, nums = [1, 2, _]` |
| `nums = [0,0,1,1,1,2,2,3,3,4]` | `5, nums = [0, 1, 2, 3, 4, _, _, _, _, _]` |
| `nums = [7]` | `1, nums = [7]` |

**🔎 How to spot it:** "**in place**", "**O(1) extra space**", "remove / keep some items", "return the new length" → a
        reader pointer that visits everything and a writer pointer that marks the next free slot.

**💡 The trick (reader and writer pointers):** `k` counts the unique values kept so far (the first value is always kept, so start at 1). The reader
        `i` walks from index 1. Because the list is **sorted**, a new value is one that differs from the **last kept**
        value `nums[k - 1]`; copy it to `nums[k]` and add 1 to `k`. Duplicates are simply skipped.

**⏱ Cost:** **O(n)** time (one walk) · **O(1)** extra space. Building a new list with a set is O(n) space and breaks the in-place rule.

```python
def remove_duplicates(nums):
    if not nums:
        return 0
    k = 1                                  # nums[:k] holds the unique values found so far
    for i in range(1, len(nums)):
        if nums[i] != nums[k - 1]:         # sorted list: a new value differs from the last one we kept
            nums[k] = nums[i]
            k += 1
    return k

for before, want in [([1, 1, 2], [1, 2]), ([0, 0, 1, 1, 1, 2, 2, 3, 3, 4], [0, 1, 2, 3, 4]),
                     ([7], [7]), ([-3, -3, -3], [-3]), ([1, 2, 3], [1, 2, 3])]:
    nums = list(before)
    k = remove_duplicates(nums)
    assert k == len(want) and nums[:k] == want, (before, k, nums)
print(remove_duplicates([0, 0, 1, 1, 1, 2, 2, 3, 3, 4]))
```

## 11 · Move Zeroes · [LeetCode 283](https://leetcode.com/problems/move-zeroes/) · 🟢 Easy

**🧸 Like you're five:** Some kids in the line are holding empty boxes (zeros). Everyone holding a toy should move to the front,
        **keeping their order**, and the empty boxes end up at the back.

**The question:** Move all `0`s in `nums` to the end while keeping the order of the other values. Do it **in place**, without copying the list.

| Input | Output |
|---|---|
| `nums = [0, 1, 0, 3, 12]` | `[1, 3, 12, 0, 0]` |
| `nums = [0]` | `[0]` |
| `nums = [1, 2]` | `[1, 2]` |

**🔎 How to spot it:** "move / push all X to the end", "keep the **relative order**", "**in place**" → reader and writer
        pointers, and **swap** instead of copy so the zeros travel back by themselves.

**💡 The trick (reader and writer pointers + swap):** `w` is the slot for the next non-zero. The reader `r` walks every slot; when `nums[r]` is not zero,
        swap it with `nums[w]` and move `w` on. Everything before `w` is non-zero and in order; the zeros are pushed
        behind. Swapping (not copying) means you never need a second loop to write the zeros.

**⏱ Cost:** **O(n)** time · **O(1)** extra space. Building a new list of non-zeros plus zeros is O(n) space and not in place.

```python
def move_zeroes(nums):
    w = 0                                        # the next slot for a non-zero value
    for r in range(len(nums)):
        if nums[r] != 0:
            nums[w], nums[r] = nums[r], nums[w]  # swap it forward; a zero (if any) goes back
            w += 1

for before, want in [([0, 1, 0, 3, 12], [1, 3, 12, 0, 0]), ([0], [0]), ([1, 2], [1, 2]),
                     ([0, 0, 1], [1, 0, 0]), ([4, -1, 0, 0, 2], [4, -1, 2, 0, 0])]:
    nums = list(before)
    move_zeroes(nums)
    assert nums == want, (before, nums)

import random
rng = random.Random(0)
for _ in range(300):
    nums = [rng.choice([0, 0, 1, 2, -3]) for _ in range(rng.randint(0, 8))]
    want = [x for x in nums if x] + [0] * nums.count(0)
    move_zeroes(nums)
    assert nums == want
nums = [0, 1, 0, 3, 12]; move_zeroes(nums); print(nums)
```

## 12 · Rotate Array · [LeetCode 189](https://leetcode.com/problems/rotate-array/) · 🟡 Medium

**🧸 Like you're five:** A train of numbered carriages goes round a loop: after 3 steps, the last 3 carriages are at the
        front. Magic way: flip the whole train around, then flip the first 3 carriages back, then flip the rest back.

**The question:** Rotate `nums` to the right by `k` steps, **in place** (`k` may be bigger than the length). Follow-up: O(1) extra space.

| Input | Output |
|---|---|
| `nums = [1,2,3,4,5,6,7], k = 3` | `[5,6,7,1,2,3,4]` |
| `nums = [-1,-100,3,99], k = 2` | `[3,99,-1,-100]` |
| `nums = [1, 2], k = 5` | `[2, 1]` |

**🔎 How to spot it:** "**rotate / shift** by k", "**in place**", "O(1) extra space" → reverse the whole thing, then
        reverse the two parts.

**💡 The trick (reverse three times):** First `k %= n` — rotating by the length changes nothing, and forgetting this is the classic bug when
        `k > n`. Then: 1) reverse the whole list, 2) reverse the first `k`, 3) reverse the remaining `n - k`. Each
        reverse is two pointers swapping toward the middle. (`nums[:] = nums[-k:] + nums[:-k]` also works, but uses
        O(n) extra space.)

**⏱ Cost:** **O(n)** time (every value is swapped about twice) · **O(1)** extra space. Moving one step at a time, k times, is O(n · k).

```python
def rotate(nums, k):
    n = len(nums)
    k %= n                                  # rotating by n steps changes nothing
    def reverse(lo, hi):                    # two pointers swap toward the middle
        while lo < hi:
            nums[lo], nums[hi] = nums[hi], nums[lo]
            lo, hi = lo + 1, hi - 1
    reverse(0, n - 1)                       # 1 2 3 4 5 6 7  ->  7 6 5 4 3 2 1
    reverse(0, k - 1)                       #                ->  5 6 7 4 3 2 1
    reverse(k, n - 1)                       #                ->  5 6 7 1 2 3 4

for before, k, want in [([1, 2, 3, 4, 5, 6, 7], 3, [5, 6, 7, 1, 2, 3, 4]), ([-1, -100, 3, 99], 2, [3, 99, -1, -100]),
                        ([1, 2], 5, [2, 1]), ([1], 0, [1]), ([1, 2, 3], 3, [1, 2, 3])]:
    nums = list(before)
    rotate(nums, k)
    assert nums == want, (before, k, nums)

import random
rng = random.Random(0)
for _ in range(300):
    nums = [rng.randint(-9, 9) for _ in range(rng.randint(1, 8))]
    k = rng.randint(0, 20)
    want = nums[-(k % len(nums)):] + nums[:-(k % len(nums))] if k % len(nums) else list(nums)
    rotate(nums, k)
    assert nums == want
nums = [1, 2, 3, 4, 5, 6, 7]; rotate(nums, 3); print(nums)
```

```python
def rotate_movie(nums, k):
    nums = list(nums)
    n = len(nums)
    k %= n
    m = Movie("Rotate Array", f"Every carriage moves {k} steps to the right, wrapping round to the front")
    m.step(f"Goal: the last **{k}** carriages end up at the front. The trick: **reverse three times**.",
           arr(nums), note(k=k))

    def reverse(lo, hi, why):
        a, b = lo, hi
        m.step(why, arr(nums, win=(a, b), ptr={"L": lo, "R": hi}))
        while lo < hi:
            nums[lo], nums[hi] = nums[hi], nums[lo]
            m.step(f"Swap `{nums[hi]}` and `{nums[lo]}`, then both ends step toward the middle.",
                   arr(nums, win=(a, b), hi={lo: "swap", hi: "swap"}, ptr={"L": lo, "R": hi}))
            lo, hi = lo + 1, hi - 1

    reverse(0, n - 1, "Step 1: reverse the **whole** train (blue = the part we are flipping).")
    reverse(0, k - 1, f"Step 2: reverse the first **{k}** carriages back into their order.")
    reverse(k, n - 1, f"Step 3: reverse the other **{n - k}** carriages back into their order.")
    m.step(f"Done: **{nums}**. Three flips, no extra list.", arr(nums, hi={x: "good" for x in range(n)}))
    return nums, m

start, k = [1, 2, 3, 4, 5, 6, 7], 3
answer, m = rotate_movie(start, k)
check = list(start); rotate(check, k)
assert answer == check == [5, 6, 7, 1, 2, 3, 4]
m.show()
```

## 13 · Valid Palindrome · [LeetCode 125](https://leetcode.com/problems/valid-palindrome/) · 🟢 Easy

**🧸 Like you're five:** A palindrome reads the same from both ends, like `racecar`. Put one finger on the first letter and one
        on the last. If they match, both fingers step in. Spaces and commas don't count — just hop over them.

**The question:** After turning uppercase letters into lowercase and **removing everything that is not a letter or a digit**, does `s` read the same forwards and backwards?

| Input | Output |
|---|---|
| `s = "A man, a plan, a canal: Panama"` | `True` |
| `s = "race a car"` | `False` |
| `s = " "` | `True` |

**🔎 How to spot it:** "**palindrome**", "reads the same **forwards and backwards**", "compare the two **ends**" → two
        pointers walking toward each other.

**💡 The trick (two pointers from both ends):** `L` starts at the left, `R` at the right. If `s[L]` is not a letter or digit, move `L` right; same for
        `R` moving left. Otherwise compare the two in lowercase: different → `False`, same → both step inward. When they
        meet, it is a palindrome. Traps: **digits count** (`"0P"` is `False`), and an empty or all-punctuation string is
        `True`.

**⏱ Cost:** **O(n)** time · **O(1)** extra space. Building a cleaned copy and comparing it with its reverse is also O(n) time but O(n) space.

```python
def is_palindrome(s):
    L, R = 0, len(s) - 1
    while L < R:
        if not s[L].isalnum():                 # hop over spaces and punctuation
            L += 1
        elif not s[R].isalnum():
            R -= 1
        elif s[L].lower() != s[R].lower():     # two real characters that differ
            return False
        else:
            L, R = L + 1, R - 1
    return True

assert is_palindrome("A man, a plan, a canal: Panama") is True
assert is_palindrome("race a car") is False
assert is_palindrome(" ") is True
assert is_palindrome("0P") is False                # digits count
assert is_palindrome(".,") is True                 # nothing left after cleaning

import random
rng = random.Random(0)
for _ in range(300):
    s = "".join(rng.choice("aAb1 ,") for _ in range(rng.randint(0, 8)))
    clean = [c.lower() for c in s if c.isalnum()]
    assert is_palindrome(s) == (clean == clean[::-1])
print(is_palindrome("A man, a plan, a canal: Panama"))
```

```python
def palindrome_movie(s):
    chars = ["␣" if ch == " " else ch for ch in s]     # draw a space so you can see it
    m = Movie("Valid Palindrome", f"Does '{s}' read the same from both ends?")
    L, R = 0, len(s) - 1
    done = {}
    m.step("One finger on each end. Hop over anything that is not a letter or digit; compare the rest, ignoring capitals.",
           arr(chars, ptr={"L": L, "R": R}))
    while L < R:
        if not s[L].isalnum():
            m.step(f"`{chars[L]}` is not a letter or digit → `L` hops over it.", arr(chars, hi={**done, L: "bad"}, ptr={"L": L, "R": R}))
            done[L] = "done"
            L += 1
        elif not s[R].isalnum():
            m.step(f"`{chars[R]}` is not a letter or digit → `R` hops over it.", arr(chars, hi={**done, R: "bad"}, ptr={"L": L, "R": R}))
            done[R] = "done"
            R -= 1
        elif s[L].lower() != s[R].lower():
            m.step(f"`{s[L]}` and `{s[R]}` are different → **not** a palindrome.", arr(chars, hi={**done, L: "bad", R: "bad"}, ptr={"L": L, "R": R}))
            return False, m
        else:
            m.step(f"`{s[L]}` and `{s[R]}` match (both are `{s[L].lower()}`) → both fingers step in.",
                   arr(chars, hi={**done, L: "good", R: "good"}, ptr={"L": L, "R": R}))
            done[L] = done[R] = "done"
            L, R = L + 1, R - 1
    m.step("The fingers met and every pair matched → **True**. 🎉",
           arr(chars, hi={k: "good" if ch.isalnum() else "done" for k, ch in enumerate(s)}))
    return True, m

answer, m = palindrome_movie("Top spot!")
assert answer == is_palindrome("Top spot!") is True
m.show()
```

## 14 · Two Sum II - Input Array Is Sorted · [LeetCode 167](https://leetcode.com/problems/two-sum-ii-input-array-is-sorted/) · 🟡 Medium

**🧸 Like you're five:** Number cards are laid out from smallest to biggest. Point at the smallest and the biggest. Too much?
        The biggest card can't be in the answer — drop it. Too little? The smallest can't be — drop it.

**The question:** `numbers` is sorted in increasing order. Return the **1-based** positions `[index1, index2]` (index1 < index2) of the two numbers that add up to `target`. Exactly one answer; use only O(1) extra space.

| Input | Output |
|---|---|
| `numbers = [2, 7, 11, 15], target = 9` | `[1, 2]` |
| `numbers = [2, 3, 4], target = 6` | `[1, 3]` |
| `numbers = [-1, 0], target = -1` | `[1, 2]` |

**🔎 How to spot it:** "**sorted**" + "find a **pair** with a sum" + "**O(1) space**" → two pointers from both ends
        (the hash map of problem 1 works too, but uses O(n) space).

**💡 The trick (two pointers from both ends):** `L` = 0, `R` = last. Look at `numbers[L] + numbers[R]`. Equal to target → done. Too small → `L += 1`
        (only a bigger left number can help). Too big → `R -= 1`. Each step throws away one number that can never be in
        the answer. The trap: the answer is **1-based**, so return `[L + 1, R + 1]`.

**⏱ Cost:** **O(n)** time (each step drops one number) · **O(1)** space. Binary search for each partner is O(n log n).

```python
def two_sum_ii(numbers, target):           # LeetCode calls it twoSum; renamed so problem 1 keeps its name
    L, R = 0, len(numbers) - 1
    while L < R:
        total = numbers[L] + numbers[R]
        if total == target:
            return [L + 1, R + 1]           # the question counts from 1
        if total < target:
            L += 1                          # too small: only a bigger left number can help
        else:
            R -= 1                          # too big: the biggest number cannot be in the answer
    return []

assert two_sum_ii([2, 7, 11, 15], 9) == [1, 2]
assert two_sum_ii([2, 3, 4], 6) == [1, 3]
assert two_sum_ii([-1, 0], -1) == [1, 2]
assert two_sum_ii([1, 2, 3, 4, 4, 9, 56, 90], 8) == [4, 5]   # equal values next to each other
print(two_sum_ii([2, 7, 11, 15], 9))
```

## 15 · 3Sum · [LeetCode 15](https://leetcode.com/problems/3sum/) · 🟡 Medium

**🧸 Like you're five:** Find three number cards that add up to zero. Sort them first. Hold one card in your hand, then play
        the two-finger game from problem 14 on the cards to its right to find the other two.

**The question:** Return all **different** triples `[a, b, c]` from `nums` (three different positions) with `a + b + c == 0`. The answer must not contain the same triple twice; order does not matter.

| Input | Output |
|---|---|
| `nums = [-1, 0, 1, 2, -1, -4]` | `[[-1, -1, 2], [-1, 0, 1]]` |
| `nums = [0, 1, 1]` | `[]` |
| `nums = [0, 0, 0]` | `[[0, 0, 0]]` |

**🔎 How to spot it:** "find **triples** (or pairs) with a sum", "**no duplicate** answers" → **sort**, fix the first
        number, and use two pointers for the other two. Sorting also makes skipping duplicates easy: equal values sit
        side by side.

**💡 The trick (sort, fix one, two pointers):** Sort. For each `i`, look for `L` and `R` to the right with `nums[L] + nums[R] == -nums[i]` using the
        two-pointer squeeze. Skip duplicates in two places: an `i` equal to the one before it, and after a hit, an `L`
        equal to the one before it. Stop early once `nums[i] > 0` — three positive numbers can't make zero. Forgetting a
        duplicate skip is the classic bug (you return `[-1, 0, 1]` twice).

**⏱ Cost:** **O(n²)** time (n fixed numbers × one O(n) squeeze; the sort is only O(n log n)) · **O(1)** extra space besides the output (sorting may use O(n)). Trying every triple is O(n³).

```python
def three_sum(nums):
    nums = sorted(nums)
    out = []
    for i in range(len(nums) - 2):
        if nums[i] > 0:                          # the smallest is positive: nothing can make 0
            break
        if i > 0 and nums[i] == nums[i - 1]:     # same first number as last time: skip, or we repeat triples
            continue
        L, R = i + 1, len(nums) - 1
        while L < R:
            total = nums[i] + nums[L] + nums[R]
            if total < 0:
                L += 1
            elif total > 0:
                R -= 1
            else:
                out.append([nums[i], nums[L], nums[R]])
                L, R = L + 1, R - 1
                while L < R and nums[L] == nums[L - 1]:   # skip a repeated middle number
                    L += 1
    return out

assert sorted(three_sum([-1, 0, 1, 2, -1, -4])) == [[-1, -1, 2], [-1, 0, 1]]
assert three_sum([0, 1, 1]) == []
assert three_sum([0, 0, 0]) == [[0, 0, 0]]
assert three_sum([0, 0, 0, 0]) == [[0, 0, 0]]            # no repeats
assert three_sum([]) == [] and three_sum([1, -1]) == []

import random
from itertools import combinations
rng = random.Random(0)
for _ in range(300):
    nums = [rng.randint(-4, 4) for _ in range(rng.randint(0, 9))]
    want = sorted(set(tuple(sorted(t)) for t in combinations(nums, 3) if sum(t) == 0))
    got = three_sum(nums)
    assert sorted(map(tuple, got)) == want and len(got) == len(want)
print(three_sum([-1, 0, 1, 2, -1, -4]))
```

```python
def three_sum_movie(nums):
    nums = sorted(nums)
    m = Movie("3Sum", "All the different triples that add up to 0")
    out = []

    def found():
        return arr([str(t) for t in out], "found", hi={k: "good" for k in range(len(out))}, index=False)

    m.step(f"Sorted: `{nums}`. Hold one number `i` (orange), then squeeze `L` and `R` to find two more that cancel it out.",
           arr(nums), found())
    for i in range(len(nums) - 2):
        if nums[i] > 0:
            m.step(f"`{nums[i]}` is positive, and everything after it is too → no zero sum is possible. Stop.",
                   arr(nums, hi={i: "bad"}, ptr={"i": i}), found())
            break
        if i > 0 and nums[i] == nums[i - 1]:
            m.step(f"`{nums[i]}` is the same as the number before it, so its triples were already found. **Skip** it.",
                   arr(nums, hi={i: "bad", i - 1: "done"}, ptr={"i": i}), found())
            continue
        L, R = i + 1, len(nums) - 1
        while L < R:
            total = nums[i] + nums[L] + nums[R]
            say_sum = f"`{nums[i]}` + `{nums[L]}` + `{nums[R]}` = {total}"
            hi = {i: "swap", L: "look", R: "look"}
            if total < 0:
                m.step(f"{say_sum}: **too small** → move `L` right to a bigger number.", arr(nums, hi=hi, ptr={"i": i, "L": L, "R": R}), found())
                L += 1
            elif total > 0:
                m.step(f"{say_sum}: **too big** → move `R` left to a smaller number.", arr(nums, hi=hi, ptr={"i": i, "L": L, "R": R}), found())
                R -= 1
            else:
                out.append([nums[i], nums[L], nums[R]])
                m.step(f"{say_sum} → a triple! Save it, then move both `L` and `R` inward.",
                       arr(nums, hi={i: "good", L: "good", R: "good"}, ptr={"i": i, "L": L, "R": R}), found())
                L, R = L + 1, R - 1
                while L < R and nums[L] == nums[L - 1]:
                    L += 1
    m.step(f"Every `i` has been tried → **{len(out)}** different triple{'s' if len(out) != 1 else ''}, no repeats.",
           arr(nums, hi={k: "done" for k in range(len(nums))}), found())
    return out, m

nums = [-1, 0, 1, 2, -1, -4]
answer, m = three_sum_movie(nums)
assert answer == three_sum(nums) == [[-1, -1, 2], [-1, 0, 1]]
m.show()
```

## 16 · Container With Most Water · [LeetCode 11](https://leetcode.com/problems/container-with-most-water/) · 🟡 Medium

**🧸 Like you're five:** Pick two walls and pour water between them. The water can only rise as high as the **shorter** wall,
        and it spreads as wide as the gap between them. Which two walls hold the most water?

**The question:** `height[i]` is the height of a wall at position `i`. Choose two walls; the water they hold is `min(height[L], height[R]) × (R − L)`. Return the largest amount.

| Input | Output |
|---|---|
| `height = [1,8,6,2,5,4,8,3,7]` | `49` |
| `height = [1, 1]` | `1` |
| `height = [4, 3, 2, 1, 4]` | `16` |

**🔎 How to spot it:** "choose **two** positions", "maximise something that depends on **the distance** and **the smaller
        of the two**" → start as wide as possible and squeeze with two pointers.

**💡 The trick (two pointers: move the shorter wall):** Start with the widest pair `L = 0`, `R = n − 1`. Work out the water, keep the best. Then move the
        **shorter** wall inward. Why? Any container that keeps the shorter wall is narrower **and** can't be taller, so
        it can never beat what we just measured — that wall is finished. Moving the taller wall is the common mistake.

**⏱ Cost:** **O(n)** time (each step drops one wall) · **O(1)** space. Trying every pair is O(n²).

```python
def max_area(height):
    L, R = 0, len(height) - 1
    best = 0
    while L < R:
        h = min(height[L], height[R])          # water stops at the shorter wall
        best = max(best, h * (R - L))
        if height[L] < height[R]:              # the shorter wall can never do better: move it
            L += 1
        else:
            R -= 1
    return best

assert max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]) == 49
assert max_area([1, 1]) == 1
assert max_area([4, 3, 2, 1, 4]) == 16
assert max_area([0, 0]) == 0

import random
rng = random.Random(0)
for _ in range(300):
    h = [rng.randint(0, 9) for _ in range(rng.randint(2, 9))]
    assert max_area(h) == max(min(h[a], h[b]) * (b - a) for a in range(len(h)) for b in range(a + 1, len(h)))
print(max_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))
```

```python
def max_area_movie(height):
    m = Movie("Container With Most Water", "Pick the two walls that hold the most water")
    L, R = 0, len(height) - 1
    best, best_pair = 0, None
    m.step("Start with the **widest** container: the two end walls. The water only rises to the **shorter** wall.",
           bars(height, ptr={"L": L, "R": R}), note(best=best))
    while L < R:
        h = min(height[L], height[R])
        area = h * (R - L)
        is_best = area > best
        if is_best:
            best, best_pair = area, (L, R)
        if height[L] < height[R]:
            move = f"`L` ({height[L]}) is shorter, so move `L`: keeping it can never beat {area}."
        elif height[L] > height[R]:
            move = f"`R` ({height[R]}) is shorter, so move `R`: keeping it can never beat {area}."
        else:
            move = "Same height — moving either is fine; we move `R`."
        m.step(f"Width {R - L} × height {h} = **{area}**{' — a new best!' if is_best else '.'} {move}",
               bars(height, hi={L: "look", R: "look"}, ptr={"L": L, "R": R}, area=(L, R, h)),
               note(area=area, best=best))
        if height[L] < height[R]:
            L += 1
        else:
            R -= 1
    a, b = best_pair
    m.step(f"The pointers met. The best was walls {a} and {b}: {min(height[a], height[b])} × {b - a} = **{best}**.",
           bars(height, hi={a: "good", b: "good"}, area=(a, b, min(height[a], height[b]))), note(best=best))
    return best, m

height = [1, 8, 6, 2, 5, 4, 8, 3, 7]
answer, m = max_area_movie(height)
assert answer == max_area(height) == 49
m.show()
```

## 17 · Sort Colors · [LeetCode 75](https://leetcode.com/problems/sort-colors/) · 🟡 Medium

**🧸 Like you're five:** A row of red (0), white (1) and blue (2) balls. Throw each red to the front pile and each blue to the
        back pile; the whites end up in the middle by themselves. One walk along the row, no counting.

**The question:** `nums` holds only 0s, 1s and 2s. Sort it **in place** so all 0s come first, then 1s, then 2s — without the library sort. Follow-up: one pass, O(1) extra space.

| Input | Output |
|---|---|
| `nums = [2, 0, 2, 1, 1, 0]` | `[0, 0, 1, 1, 2, 2]` |
| `nums = [2, 0, 1]` | `[0, 1, 2]` |
| `nums = [1]` | `[1]` |

**🔎 How to spot it:** "only **three kinds** of values", "group them", "**one pass**, in place" → three pointers: `low`
        (end of the 0s), `mid` (the checker), `high` (start of the 2s).

**💡 The trick (three pointers (dutch flag)):** Keep: everything before `low` is 0, between `low` and `mid` is 1, after `high` is 2; `mid..high` is
        not checked yet. Look at `nums[mid]`: a 0 → swap with `low`, move both `low` and `mid`; a 1 → just move `mid`;
        a 2 → swap with `high` and move `high` back, but **do not move `mid`** — the value that came from the back has
        not been checked yet. That last rule is the one everybody gets wrong.

**⏱ Cost:** **O(n)** time (one pass) · **O(1)** space. Counting the 0s, 1s and 2s and rewriting is also O(n) but takes two passes.

```python
def sort_colors(nums):
    low, mid, high = 0, 0, len(nums) - 1     # [0, low) = 0s, [low, mid) = 1s, (high, end] = 2s
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low, mid = low + 1, mid + 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1                         # do NOT move mid: the value from the back is not checked yet

for before in [[2, 0, 2, 1, 1, 0], [2, 0, 1], [1], [], [2, 2, 0, 0], [0, 1, 2, 0, 1, 2]]:
    nums = list(before)
    sort_colors(nums)
    assert nums == sorted(before), (before, nums)

import random
rng = random.Random(0)
for _ in range(300):
    nums = [rng.randint(0, 2) for _ in range(rng.randint(0, 10))]
    want = sorted(nums)
    sort_colors(nums)
    assert nums == want
nums = [2, 0, 2, 1, 1, 0]; sort_colors(nums); print(nums)
```

```python
def sort_colors_movie(nums):
    nums = list(nums)
    n = len(nums)
    m = Movie("Sort Colors", "Reds (0) to the front, blues (2) to the back, whites (1) in the middle")
    low, mid, high = 0, 0, n - 1

    def pic(hi):
        settled = {k: "done" for k in range(n) if k < low or k > high}
        return arr(nums, hi={**settled, **hi}, ptr={"low": low, "mid": mid, "high": high})

    m.step("Left of `low` = 0s, right of `high` = 2s (grey = settled). `mid` checks the unknown middle, one ball at a time.", pic({}))
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            m.step("`mid` finds a **0** → swap it to `low` at the front. Both `low` and `mid` step right.",
                   pic({low: "swap", mid: "swap"}))
            low, mid = low + 1, mid + 1
        elif nums[mid] == 1:
            m.step("`mid` finds a **1** → it already belongs in the middle. Just move `mid` on.", pic({mid: "look"}))
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            m.step(f"`mid` finds a **2** → swap it to `high` at the back. `high` steps left, but `mid` stays: the `{nums[mid]}` that came back is not checked yet!",
                   pic({mid: "swap", high: "swap"}))
            high -= 1
    m.step(f"`mid` passed `high`: nothing is left to check → **{nums}**.", arr(nums, hi={k: "good" for k in range(n)}))
    return nums, m

start = [2, 0, 2, 1, 1, 0]
answer, m = sort_colors_movie(start)
check = list(start); sort_colors(check)
assert answer == check == [0, 0, 1, 1, 2, 2]
m.show()
```

## 18 · Subarray Sum Equals K · [LeetCode 560](https://leetcode.com/problems/subarray-sum-equals-k/) · 🟡 Medium

**🧸 Like you're five:** You walk along a path of numbered stones, keeping a running total in your head, and you write every
        total in a notebook. If your total now minus some total you had earlier is exactly `k`, the stones between
        those two moments add up to `k`.

**The question:** Given `nums` (it may contain **negative** numbers) and `k`, return how many contiguous, non-empty subarrays add up to exactly `k`.

| Input | Output |
|---|---|
| `nums = [1, 1, 1], k = 2` | `2` |
| `nums = [1, 2, 3], k = 3` | `2` |
| `nums = [1, -1, 0], k = 0` | `3` |

**🔎 How to spot it:** "**subarray sum equals k**" / "count the subarrays with sum …", especially with **negative numbers**
        → prefix sums + a dictionary of how often each prefix sum happened. (A sliding window does **not** work with
        negatives: adding a number can make the sum smaller.)

**💡 The trick (prefix sums + hash map):** Keep a running total. A subarray ending here sums to `k` exactly when some **earlier** running total
        equals `total − k`. So keep a dictionary `seen[prefix] = how many times`, add `seen[total − k]` to the answer
        at every step, then record `total`. Start with `seen = {0: 1}` — the empty prefix — or you miss subarrays
        that start at index 0. That missing `{0: 1}` is the classic bug.

**⏱ Cost:** **O(n)** time (one walk, O(1) lookups) · **O(n)** space for the dictionary. Checking every subarray is O(n²).

```python
def subarray_sum(nums, k):
    seen = {0: 1}                 # running total -> how many times we've had it; the empty start counts once
    total = count = 0
    for x in nums:
        total += x
        count += seen.get(total - k, 0)    # each earlier total of (total - k) starts a subarray summing to k
        seen[total] = seen.get(total, 0) + 1
    return count

assert subarray_sum([1, 1, 1], 2) == 2
assert subarray_sum([1, 2, 3], 3) == 2
assert subarray_sum([1, -1, 0], 0) == 3               # negatives and zeros
assert subarray_sum([5], 5) == 1 and subarray_sum([5], 4) == 0

import random
rng = random.Random(0)
for _ in range(300):
    nums = [rng.randint(-3, 3) for _ in range(rng.randint(1, 9))]
    k = rng.randint(-3, 4)
    brute = sum(sum(nums[a:b]) == k for a in range(len(nums)) for b in range(a + 1, len(nums) + 1))
    assert subarray_sum(nums, k) == brute
print(subarray_sum([1, 1, 1], 2))
```

```python
def subarray_sum_movie(nums, k):
    m = Movie("Subarray Sum Equals K", f"How many unbroken stretches add up to {k}?")
    seen = {0: 1}
    where = {0: [-1]}                    # only for the picture: where each running total happened
    total = count = 0
    title = "notebook  (running total → times)"
    m.step(f"A stretch adds to {k} when **total now − total earlier = {k}**. So at each stone, look up `total − {k}` in the notebook.",
           arr(nums), hmap(seen, title), note(total=total, count=count))
    for i, x in enumerate(nums):
        total += x
        need = total - k
        hits = seen.get(need, 0)
        count += hits
        if hits:
            j = where[need][-1]
            cap = f"Total is `{total}`. `{total} − {k} = {need}` IS in the notebook → the green stretch `{nums[j + 1:i + 1]}` adds to {k}. Count = **{count}**."
            hi = {**{t: "good" for t in range(j + 1, i + 1)}, i: "good"}
            mhi = {need: "good"}
        else:
            cap = f"Add `{x}`: total is `{total}`. Is `{need}` in the notebook? No, so no stretch ending here adds to {k}."
            hi = {**{t: "done" for t in range(i)}, i: "look"}
            mhi = {}
        seen[total] = seen.get(total, 0) + 1
        where.setdefault(total, []).append(i)
        m.step(cap, arr(nums, hi=hi, ptr={"i": i}), hmap(seen, title, hi={**mhi, total: mhi.get(total, "new")}),
               note(total=total, need=need, count=count))
    m.step(f"End of the path: **{count}** stretches add up to {k}, found in one walk.",
           arr(nums, hi={t: "done" for t in range(len(nums))}), hmap(seen, title), note(count=count))
    return count, m

nums, k = [3, 4, 7, 2, -3, 1, 4, 2], 7
answer, m = subarray_sum_movie(nums, k)
assert answer == subarray_sum(nums, k) == 4
m.show()
```

## 19 · Valid Sudoku · [LeetCode 36](https://leetcode.com/problems/valid-sudoku/) · 🟡 Medium

**🧸 Like you're five:** A Sudoku rule-checker keeps 27 little bags: one for every row, every column and every 3×3 box. For
        each digit on the board, it peeks in that digit's three bags. Already there? Rule broken!

**The question:** Given a partly filled 9×9 board (`"."` = empty), return `True` if the **filled** cells break no rule: no digit twice in a row, in a column, or in one of the nine 3×3 boxes. The board does not have to be solvable.

| Input | Output |
|---|---|
| `board = LeetCode example 1` | `True` |
| `same board, but top-left 5 → 8` | `False (two 8s in the top-left box)` |
| `an empty board` | `True` |

**🔎 How to spot it:** "**no repeats** within each group" where a cell belongs to several groups → one **set per group**
        and a single pass over the cells.

**💡 The trick (one set per row, column and box):** Make 9 row sets, 9 column sets and 9 box sets. For every filled cell `(r, c)` with digit `v`, its box
        is `(r // 3) * 3 + c // 3` (boxes numbered 0–8, left to right, top to bottom). If `v` is already in the row,
        column or box set → `False`. Otherwise add it to all three. Getting the box number formula wrong is the
        usual bug; test it on the four corners.

**⏱ Cost:** **O(1)** time and space — the board is always 81 cells (O(n²) for an n×n board). Checking every row, then every column, then every box separately is the same big-O, just three passes.

```python
def is_valid_sudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]
    for r in range(9):
        for c in range(9):
            v = board[r][c]
            if v == ".":
                continue
            b = (r // 3) * 3 + c // 3              # which 3x3 box: 0..8, left to right, top to bottom
            if v in rows[r] or v in cols[c] or v in boxes[b]:
                return False
            rows[r].add(v); cols[c].add(v); boxes[b].add(v)
    return True

EXAMPLE_1 = [list(row) for row in ["53..7....", "6..195...", ".98....6.", "8...6...3", "4..8.3..1",
                                    "7...2...6", ".6....28.", "...419..5", "....8..79"]]
EXAMPLE_2 = [row[:] for row in EXAMPLE_1]
EXAMPLE_2[0][0] = "8"                              # now two 8s share the top-left box
assert is_valid_sudoku(EXAMPLE_1) is True
assert is_valid_sudoku(EXAMPLE_2) is False
assert is_valid_sudoku([["."] * 9 for _ in range(9)]) is True
same_col = [["."] * 9 for _ in range(9)]; same_col[0][4] = same_col[8][4] = "3"
assert is_valid_sudoku(same_col) is False          # a column clash in different boxes
same_row = [["."] * 9 for _ in range(9)]; same_row[5][0] = same_row[5][8] = "9"
assert is_valid_sudoku(same_row) is False          # a row clash in different boxes
assert [(r // 3) * 3 + c // 3 for r, c in [(0, 0), (0, 8), (8, 0), (8, 8), (4, 4)]] == [0, 2, 6, 8, 4]
print(is_valid_sudoku(EXAMPLE_1), is_valid_sudoku(EXAMPLE_2))
```

```python
def sudoku_movie(board):
    m = Movie("Valid Sudoku", "No digit twice in a row, a column or a 3×3 box")
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]
    view = [["" if v == "." else v for v in row] for row in board]
    checked = {}

    def pic(r, c, b, extra):
        top, left = (b // 3) * 3, (b % 3) * 3
        hi = {(rr, cc): "win" for rr in range(top, top + 3) for cc in range(left, left + 3)}   # blue = this cell's box
        hi.update(checked)
        hi.update(extra)
        return grid(view, hi=hi, row_labels=range(9), col_labels=range(9))

    def bags(r, c, b, v, role):
        return (hmap(rows[r], f"row {r} has", hi={v: role}), hmap(cols[c], f"column {c} has", hi={v: role}),
                hmap(boxes[b], f"box {b} has", hi={v: role}))

    m.step("Check the filled cells one by one, row by row. Each digit must be new in its **row**, its **column** and its **box** (blue).",
           grid(view, row_labels=range(9), col_labels=range(9)))
    for r in range(9):
        for c in range(9):
            v = board[r][c]
            if v == ".":
                continue
            b = (r // 3) * 3 + c // 3
            if v in rows[r] or v in cols[c] or v in boxes[b]:
                where = "row" if v in rows[r] else "column" if v in cols[c] else "box"
                twin = next((rr, cc) for rr in range(9) for cc in range(9) if (rr, cc) in checked and board[rr][cc] == v
                            and (rr == r if where == "row" else cc == c if where == "column" else (rr // 3) * 3 + cc // 3 == b))
                m.step(f"`{v}` at row {r}, column {c}: the **{where}** bag already has a `{v}` (at row {twin[0]}, column {twin[1]}). Rule broken → **False**.",
                       pic(r, c, b, {(r, c): "bad", twin: "bad"}), *bags(r, c, b, v, "bad"))
                return False, m
            rows[r].add(v); cols[c].add(v); boxes[b].add(v)
            m.step(f"`{v}` at row {r}, column {c} (box {b}): not in any of its three bags yet → put it in all three.",
                   pic(r, c, b, {(r, c): "look"}), *bags(r, c, b, v, "new"))
            checked[(r, c)] = "done"
    m.step("Every filled cell passed its three checks → **True**.", grid(view, hi={rc: "good" for rc in checked}))
    return True, m

answer, m = sudoku_movie(EXAMPLE_2)
assert answer == is_valid_sudoku(EXAMPLE_2) is False
m.show()
```

## 20 · Longest Common Prefix · [LeetCode 14](https://leetcode.com/problems/longest-common-prefix/) · 🟢 Easy

**🧸 Like you're five:** Line the words up, one under the other, like a stack of name tags. Read down the first column: all the
        same letter? Keep it and go to the next column. The first column where a letter differs — or a word runs out —
        is where you stop.

**The question:** Return the longest string that is a prefix (a beginning) of **every** string in `strs`, or `""` if there is none.

| Input | Output |
|---|---|
| `strs = ["flower","flow","flight"]` | `"fl"` |
| `strs = ["dog","racecar","car"]` | `""` |
| `strs = ["a"]` | `"a"` |

**🔎 How to spot it:** "**common prefix** / start shared by all" → compare the words **column by column** (vertical
        scanning), stopping at the first mismatch. (For *many* prefix questions over the same words, build a trie —
        part 4.)

**💡 The trick (compare column by column):** Take the first word as the ruler. For each position `i` of it, check every other word: if the word is
        too short (`i == len(w)`) or has a different letter, the answer is `first[:i]`. If the loop finishes, the whole
        first word is the answer. The trap: forgetting the *too short* check and reading past the end of a word.

**⏱ Cost:** **O(S)** time, where S is the total number of letters (we stop at the first mismatch) · **O(1)** extra space. Sorting the words and comparing only the first and last also works, in O(S log n).

```python
def longest_common_prefix(strs):
    if not strs:
        return ""
    first = strs[0]
    for i, ch in enumerate(first):           # try each column of letters
        for w in strs[1:]:
            if i == len(w) or w[i] != ch:    # a word ran out, or a letter differs
                return first[:i]
    return first

assert longest_common_prefix(["flower", "flow", "flight"]) == "fl"
assert longest_common_prefix(["dog", "racecar", "car"]) == ""
assert longest_common_prefix(["a"]) == "a"
assert longest_common_prefix(["", "b"]) == ""
assert longest_common_prefix(["ab", "a"]) == "a"        # a shorter word later on

import os, random
rng = random.Random(0)
for _ in range(300):
    strs = ["".join(rng.choice("ab") for _ in range(rng.randint(0, 4))) for _ in range(rng.randint(1, 4))]
    assert longest_common_prefix(strs) == os.path.commonprefix(strs)
print(repr(longest_common_prefix(["flower", "flow", "flight"])))
```

```python
def lcp_movie(strs):
    m = Movie("Longest Common Prefix", "Stack the words and read down the columns")
    width = max(len(w) for w in strs)
    rows = [list(w) + [""] * (width - len(w)) for w in strs]     # pad short words with empty boxes

    def pic(i, extra):
        hi = {(r, c): "good" for r in range(len(strs)) for c in range(i)}   # columns everyone shares
        hi.update(extra)
        return grid(rows, hi=hi, row_labels=range(len(strs)), col_labels=range(width))

    first = strs[0]
    m.step(f"Word 0, `{first}`, is the ruler. Read down each column: while every word has the same letter, the prefix grows.",
           pic(0, {}), note(prefix="''"))
    for i, ch in enumerate(first):
        for r in range(1, len(strs)):
            w = strs[r]
            if i == len(w) or w[i] != ch:
                why = f"`{w}` has no letter at position {i}" if i == len(w) else f"`{w}` has `{w[i]}` in column {i}, not `{ch}`"
                m.step(f"{why} → stop here.", pic(i, {(0, i): "look", (r, i): "bad"}), note(prefix=repr(first[:i])))
                m.step(f"Everything left of column {i} is shared by all the words → the answer is **{first[:i]!r}**.",
                       pic(i, {}), note(prefix=repr(first[:i])))
                return first[:i], m
            m.step(f"Column {i}: `{w}` also has `{ch}`. ✓", pic(i, {(0, i): "look", (r, i): "good"}), note(prefix=repr(first[:i])))
    m.step(f"The whole first word is shared → **{first!r}**.", pic(len(first), {}), note(prefix=repr(first)))
    return first, m

strs = ["flower", "flow", "flight"]
answer, m = lcp_movie(strs)
assert answer == longest_common_prefix(strs) == "fl"
m.show()
```

## Quiz yourself: which trick?

These eight are **not** in the top 100, so you have never seen them here. Read each one, name the trick out loud,
then open the answer.

1. **Squares of a Sorted Array** ([LeetCode 977](https://leetcode.com/problems/squares-of-a-sorted-array/), Easy) — `nums` is sorted and may contain negatives; return the squares, also sorted, in O(n).
   <details><summary>which trick?</summary>Two pointers from both ends: the biggest square is always at one of the two ends (a big negative or a big positive), so fill the answer from the back.</details>
2. **Isomorphic Strings** ([LeetCode 205](https://leetcode.com/problems/isomorphic-strings/), Easy) — can you replace letters of `s` one-for-one to get `t`?
   <details><summary>which trick?</summary>Hash maps: remember <code>s-letter → t-letter</code> and <code>t-letter → s-letter</code>; a letter that ever maps two ways breaks it.</details>
3. **Intersection of Two Arrays** ([LeetCode 349](https://leetcode.com/problems/intersection-of-two-arrays/), Easy) — the values that appear in both lists, each once.
   <details><summary>which trick?</summary>Hash set: put one list in a set and keep the values of the other that are in it (or <code>set(a) &amp; set(b)</code>).</details>
4. **Continuous Subarray Sum** ([LeetCode 523](https://leetcode.com/problems/continuous-subarray-sum/), Medium) — is there a stretch of length ≥ 2 whose sum is a multiple of `k`?
   <details><summary>which trick?</summary>Prefix sums + hash map, like problem 18: store the first index of each <code>total % k</code>; the same remainder twice, at least 2 apart, means the stretch between is a multiple of k.</details>
5. **Remove Element** ([LeetCode 27](https://leetcode.com/problems/remove-element/), Easy) — remove every copy of `val` in place and return how many values are left.
   <details><summary>which trick?</summary>Reader and writer pointers, like problems 10 and 11: copy each value that is not <code>val</code> to the writer's slot.</details>
6. **4Sum** ([LeetCode 18](https://leetcode.com/problems/4sum/), Medium) — all different quadruples adding up to `target`.
   <details><summary>which trick?</summary>3Sum with one more loop: sort, fix two numbers, squeeze two pointers for the other two, and skip duplicates at every level. O(n³).</details>
7. **Boats to Save People** ([LeetCode 881](https://leetcode.com/problems/boats-to-save-people/), Medium) — each boat carries at most two people and at most `limit` weight; fewest boats?
   <details><summary>which trick?</summary>Sort + two pointers from both ends: the heaviest person always leaves on the next boat; the lightest joins them if they fit.</details>
8. **Find All Duplicates in an Array** ([LeetCode 442](https://leetcode.com/problems/find-all-duplicates-in-an-array/), Medium) — values are 1..n, some appear twice; find them in O(n) time.
   <details><summary>which trick?</summary>"Seen before?" → a hash set. For the O(1)-space follow-up, use the list itself as the notebook: flip <code>nums[abs(x) - 1]</code> negative the first time you see x; if it is already negative, x is a duplicate.</details>

## Say it in an interview

- **Clarify first.** Is the list sorted? Can there be negatives, zeros, duplicates? May I change the input? Is
  there always an answer? Indices or values — and counting from 0 or 1 (Two Sum II is 1-based)?
- **Brute force out loud, then name the repeated work.** *"Trying every pair is O(n²). The slow part is searching
  for the partner again and again — a dictionary makes that O(1), so O(n) overall."*
- **Name the trick and its promise.** *"The list is sorted, so two pointers: every step throws away one number
  that can't be in the answer."* Saying why a pointer may move is what convinces the interviewer.
- **Give time and space, and the trade-off.** Hash set: O(n) time, O(n) space. Sort first: O(n log n) time, O(1)
  extra space. Offer both when the question does not say which matters.
- **Test out loud:** empty list, one item, all the same, negatives, zeros, the answer at the very start or end,
  `k` bigger than `n`.
- **Common traps in this part:** storing `x` before checking for its partner (it pairs with itself); forgetting
  `seen = {0: 1}` in prefix sums; using a sliding window when there are negatives; forgetting `k %= n`; returning
  the same triple twice in 3Sum; moving `mid` after swapping with `high` in Sort Colors; moving the *taller* wall
  in Container With Most Water.
